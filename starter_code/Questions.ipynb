### a: If Alice and Bob Never Update Their DH Key (Break-in Recovery and Forward Secrecy)

- **Break-in Recovery**:
  - Without DH updates, no new randomness is introduced.
  - Compromised keys allow attackers to decrypt all future messages.
  - **Result**: No break-in recovery.
- **Forward Secrecy**:
  - Symmetric ratchet (`kdf_ck`) derives unique message keys, updated via one-way HMAC.
  - Past messages remain secure if keys are compromised later.
  - **Result**: Forward secrecy maintained.

### c: Security Feature Preventing Access to Locker Combination

- **Feature**: Forward Secrecy.
- **How Double Ratchet Provides It**:
  - Symmetric ratchet (`kdf_ck`) derives unique message key for Message 3 ("Sure, it's 1234") using HMAC-SHA256.
  - Chain key updates after each message, making prior message keys non-derivable.
  - Attacker with keys before Message 3 cannot decrypt it, as the message key is lost after `send_ck` updates.

### d: RSA and ECDSA Comparison

### ECDSA
ECDSA, implemented in `question_4_code.py` using the P-384 curve, is highly efficient across all operations. **Key generation** is fast, taking approximately 2-20 ms (2.35 ms in tests), as generating elliptic curve key pairs is computationally lightweight (`lib.py`, line 258). **Signature generation** is quick, around 3-10 ms (3.24 ms in tests), involving a SHA384 hash and elliptic curve signing (`lib.py`, lines 271-275). The **signature length** is compact at 96 bytes, fixed for the P-384 curve, as seen in the test output. **Verification** is moderately fast, typically 3-15 ms (3.58 ms in tests), requiring a hash and elliptic curve operations (`lib.py`, lines 48-60). ECDSA’s efficiency makes it ideal for resource-constrained systems.

### RSA
RSA, implemented in `question_4_code.py` with a 4096-bit key, is computationally heavier than ECDSA. **Key generation** is slow, ranging from 500-3000 ms (680.22 ms in tests), due to the intensive process of finding large primes (`question_4_code.py`, lines 44-45). **Signature generation** takes 8-100 ms (8.01 ms in tests), as it involves SHA256 hashing and PSS signing with a large key (`question_4_code.py`, lines 50-52). The **signature length** is fixed at 512 bytes, matching the key size, as confirmed in tests. **Verification** is faster, around 1-10 ms (1.17 ms in tests), using a single public-key operation (`question_4_code.py`, lines 55-62). RSA’s performance is slower, but its verification speed is a strength.
