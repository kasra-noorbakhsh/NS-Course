### a: If Alice and Bob Never Update Their DH Key (Break-in Recovery and Forward Secrecy)

- **Break-in Recovery**:
  - Without DH updates, no new randomness is introduced.
  - Compromised keys allow attackers to decrypt all future messages.
  - **Result**: No break-in recovery.
- **Forward Secrecy**:
  - Symmetric ratchet (`kdf_ck`) derives unique message keys, updated via one-way HMAC.
  - Past messages remain secure if keys are compromised later.
  - **Result**: Forward secrecy maintained.

### b: Longest Sending Chain for Alice and Bob

- **Conversation**:
  1. A: "Hey Bob, can you send me the locker combo" (Alice chain: 1)
  2. A: "I need to get my laptop" (Alice chain: 2)
  3. B: "Sure, it's 1234" (Bob chain: 1)
  4. A: "Great, thanks!..." (Alice chain: 1)
  5. B: "Did it work" (Bob chain: 1)
- **Longest Sending Chain**:
  - **Alice**: 2 messages (Messages 1-2).
  - **Bob**: 1 message (no consecutive sends).

### c: Security Feature Preventing Access to Locker Combination

- **Feature**: Forward Secrecy.
- **How Double Ratchet Provides It**:
  - Symmetric ratchet (`kdf_ck`) derives unique message key for Message 3 ("Sure, it's 1234") using HMAC-SHA256.
  - Chain key updates after each message, making prior message keys non-derivable.
  - Attacker with keys before Message 3 cannot decrypt it, as the message key is lost after `send_ck` updates.

### d: RSA and ECDSA Comparission

- **Key Generation Time**: RSA requires significantly more time (~2000-3000 ms) than ECDSA (~10-20 ms) due to the complexity of generating large primes versus elliptic curve points.

- **Signature Generation Time**: RSA is slower (~50-100 ms) than ECDSA (~5-10 ms) because of the larger key size and PSS computations compared to ECDSA’s efficient curve operations.

- **Signature Length**: RSA signatures are longer (512 bytes) than ECDSA signatures (~96 bytes), as RSA signature size scales with the key length, while ECDSA produces compact, fixed-size signatures.

- **Verification Time**: ECDSA verification (~10-15 ms) is slightly slower than RSA (~5-10 ms), as ECDSA involves elliptic curve operations, while RSA verification uses a single, faster public-key operation.
